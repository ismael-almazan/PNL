```python
# Librerias requeridos
!python -m spacy download es_core_news_sm -q
import spacy
import seaborn as sns
import matplotlib.pyplot as plt
from collections import Counter
from spacy.glossary import GLOSSARY
```

```python
texto = "En los últimos años, muchas universidades han incorporado herramientas digitales para apoyar el aprendizaje de los estudiantes. Estas tecnologías permiten organizar materiales, facilitar la comunicación y ofrecer actividades interactivas dentro y fuera del aula. Sin embargo, su uso no siempre garantiza mejores resultados académicos, porque también depende de la planeación del docente y del compromiso del alumno. En algunos cursos, los estudiantes participan más cuando utilizan plataformas virtuales, foros o sistemas de retroalimentación inmediata. Además, los profesores pueden analizar con mayor facilidad el avance del grupo y detectar dificultades en temas específicos. Por esta razón, el uso de recursos tecnológicos se ha convertido en un tema relevante para la educación actual."
```

# Tekenizar

Para separa las palabras se utilizara la libreria de spacy

```python
nlp = spacy.load("es_core_news_sm")

# Toquenizar
doc = nlp(texto)
tokens = [token.text for token in doc]
print("Tokens", tokens)
```

# Etiquetado

EL eiquetado gramatical se implementara de igual forma mediante la libreria de spacy, el cual permite un etiquetado POS medinate redes neuronales convolucionales.

```python
# Etiquetado gramatical con spacy
etiquetas = [token.pos_ for token in doc]

# Mostrar los tokens y su respectiva etiqueta
for token, etiqueta in zip(tokens, etiquetas):
    print(token, etiqueta)

# El texto contiene puntos y comas los cuales tambien etiqueto, por lo que es conveniente eliminarlos para la representacion grafica
tokens = [token.text for token in doc if token.pos_ != "PUNCT"]
etiquetas = [token.pos_ for token in doc if token.pos_ != "PUNCT"]

# Imprimir
for token, etiqueta in zip(tokens, etiquetas):
    print(token, etiqueta)
```

# Analisis  exploratorio de datos
a)	 el número total de tokens

b)  número de tokens distintos


c.	conteo de cada categoría gramatical en una gráfica de barras

d.	los cinco tokens más frecuentes del texto con sus respectivas etiquetas gramaticales en una gráfica de barras

e.	las tres clases gramaticales más frecuentes del texto en una gráfica de barras


```python
# Numero de tokens
print("Numero de tokens:", len(tokens))

# NUmero de tonks distintos
print("Numero de tokens distintos:", len(set(tokens)))

print("\n")
# Grafica de barras con categorias gramaticas con seaborn
import seaborn as sns
sns.countplot(x=etiquetas, palette="viridis")
plt.title("Conteo de categorias gramaticales")
plt.xlabel("Etiquetas")
plt.ylabel("Frecuencia")
plt.show()

 # Los cinco tokens más frecuentes del texto con sus respectivas etiquetas gramaticales en una gráfica de barras
top5 = Counter(tokens).most_common(5)
top5_tokens = [t for t, _ in top5]
top5_frec = [f for _, f in top5]
top5_etiquetas = [etiquetas[tokens.index(t)] for t in top5_tokens]
eje_x = [f"{t}\n({e})" for t, e in zip(top5_tokens, top5_etiquetas)]

sns.barplot(x=eje_x, y=top5_frec, palette="viridis")
plt.title("Top 5 Tokens")
plt.xlabel("Etiqueta")
plt.ylabel("Frecuencia")
plt.show()

print("\n")
# Llas tres clases gramaticales más frecuentes del texto en una gráfica de barras
top3 = Counter(etiquetas).most_common(3)
top3_cats = [c for c, _ in top3]
top3_frec = [f for _, f in top3]

sns.barplot(x=top3_cats, y=top3_frec, palette="viridis")
plt.title("Top 3 Etiquetas Gramaticales")
plt.xlabel("Token y Etiqueta")
plt.ylabel("Frecuencia")
plt.show()
```

# Conclusion

Al analizar el texto con etiquetado gramatical usando spaCy, se puede ver que el texto esta formado por 111 tokens, de los cuales 89 son diferentes entre si (estos quitando los puntos y comas) . Esto nos dice que el texto  no repite mucho las mismas palabras, lo que hace que el texto se sienta variado y fluido.


Lo mas interesante es que los sustantivos son las palabras que mas aparecen, lo cual tiene mucho sentido porque el texto habla de cosas concretas como tecnologias, universidades, estudiantes y aprendizaje. Despues le siguen los verbos y los determinantes.

Las palabras que mas se repiten son articulos y preposiciones como "de", "los", "el" "de", "y", que son palabras cortas pero necesarias para unir todas las ideas del texto (stopwords).

En conclusion, el etiquetado gramatical nos ayuda a entender mejor como esta construido un texto. En este caso pudimos confirmar que se trata de un texto formal y educativo, donde predominan los Sustantivos, verbos, Adposicion/Preposicion y Determinantes.
